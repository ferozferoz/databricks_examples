```python
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
dataset_name = dbutils.widgets.get("dataset_name")
run_id = dbutils.widgets.get("run_id")
partition_by = dbutils.widgets.get("partition_by")

print(f"dataset_name={dataset_name}, run_id={run_id}")

#files = dbutils.fs.ls(f"/databricks-datasets/{table_name}/{partition}")
```

```python
df = spark.read.csv(
    f"/databricks-datasets/{dataset_name}/{run_id}",
    header=True,
    inferSchema=True,
)
```

```python
from pyspark.sql.functions import to_date, col

df = df.withColumn("cobdate", to_date(col("InvoiceDate"), "M/d/yy H:mm"))
    
df.show()
```

```python
df.write.format("delta")\
.mode("overwrite")\
.partitionBy(partition_by)\
.saveAsTable("databricks_example.ambreen_nasir178.online_retail_enriched")
```
